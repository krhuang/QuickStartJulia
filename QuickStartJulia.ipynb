```julia
# import the PicoSAT package (see https://juliapackages.com/p/picosat) and generate some random cnf formulas
# How large of a formula is feasible to solve, on your machine?

using PicoSAT
using Random

function random_cnf(nvars::Int, nclauses::Int, k::Int; seed=nothing)
    if seed !== nothing
        Random.seed!(seed)
    end

    cnf = Vector{Vector{Int}}(undef, nclauses)
    for i in 1:nclauses
        vars = rand(1:nvars, k)                  # pick k variables
        signs = rand(Bool, k)                    # choose negations
        clause = [signs[j] ? vars[j] : -vars[j] for j in 1:k]
        cnf[i] = clause
    end
    return cnf
end

```

```julia
# Consider the following code for computing fibonacci numbers. 
# What are your impressions? Try calculating some large fibonacci numbers
function fib(n::Int64) 
    if n < 0
        throw("invalid input")
    elseif n == 0
        return 0
    elseif n == 1
        return 1
    else
        return fib(n-1) + fib(n-2)
    end
end

fib(-1)
```
